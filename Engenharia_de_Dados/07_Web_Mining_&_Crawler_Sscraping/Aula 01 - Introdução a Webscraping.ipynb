## Conceitos-chave

- __Web scraping__ é a prática de coletar dados diretamente de páginas web de forma automatizada.  
- Para engenheiros de dados, ele se torna uma **fonte alternativa** de ingestão quando APIs ou dumps não estão disponíveis ou não atendem aos requisitos de negócio.  
- Comparação rápida:
  - **APIs**: contrato estável, limites conhecidos, ideais para produção.  
  - **Dumps**: dados em lote, baixa frequência de atualização.  
  - **Scraping**: flexível e imediato, mas mais frágil a mudanças no layout e sujeito a restrições legais.  
- Papel do engenheiro: transformar scraping em **pipeline ETL/ELT reprodutível**, integrado ao ecossistema de dados.  

## Contexto em Engenharia de Dados
- **Ingestão**: páginas → camada bronze (HTML bruto + metadados).  
- **Parsing e normalização**: extração e padronização de campos.  
- **Armazenamento**: Parquet/Delta em data lake ou persistência em bancos.  
- **Consumo**: Spark, dashboards e sistemas de streaming.  
- **Observabilidade**: logs e métricas para acompanhar quebras de layout.  
- **Orquestração**: Airflow, Prefect ou jobs agendados.  

## Boas práticas e compliance
- Respeitar termos de uso e robots.txt.  
- Evitar coleta de dados pessoais desnecessários.  
- Documentar fontes, periodicidade e possíveis vieses.  
- Usar logging estruturado para auditoria.  

## Bibliotecas utilizadas
- **requests** → coletar HTML e XML de forma programática.  
- **logging** → monitoramento básico e registro de falhas.
