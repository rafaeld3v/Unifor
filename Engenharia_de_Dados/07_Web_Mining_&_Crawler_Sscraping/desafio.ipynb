```python
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

URL = "https://www.scrapingcourse.com/login/cf-antibot"
EMAIL = "admin@example.com"
PASSWORD = "password"

driver = uc.Chrome(headless=False, version_main=140)
driver.get(URL)

wait = WebDriverWait(driver, 45)
login_button_selector = (By.CSS_SELECTOR, "button[type='submit']")
wait.until(EC.element_to_be_clickable(login_button_selector))

driver.find_element(By.NAME, "email").send_keys(EMAIL)
driver.find_element(By.NAME, "password").send_keys(PASSWORD)
driver.find_element(By.CSS_SELECTOR, "button[type='submit']").click()

time.sleep(30)
driver.quit()
```
